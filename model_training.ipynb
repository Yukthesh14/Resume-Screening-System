```python
import pandas as pd


# ============================================================
# 1. SHORTLIST RESUMES
# ============================================================

shortlist_resumes = [

    "Data scientist with Python SQL machine learning and real world projects",

    "Python developer with 2 years experience in APIs deployment and cloud",

    "Data analyst with SQL Python analytics and dashboard development experience",

    "Machine learning engineer with Python deep learning and model deployment",

    "Software developer with Python SQL and cloud experience",

    "Data science intern with Python SQL machine learning and strong projects",

    "Python developer with internship experience and several completed projects",

    "Business analyst with SQL analytics and strong stakeholder experience",

    "Machine learning graduate with Python SQL and practical projects",

    "Cloud engineer with Python automation and deployment experience",

    "Data analyst with three years experience in SQL Excel Python and reporting",

    "Junior data scientist with Python pandas SQL and machine learning projects",

    "Python backend developer with Django REST APIs SQL and cloud deployment",

    "Machine learning intern with Python scikit learn and real world project experience",

    "Data analyst experienced in SQL Python visualization and business reporting",

    "Software engineer with Python APIs database management and cloud experience",

    "Data science graduate with Python SQL statistics and practical machine learning projects",

    "Python developer with one year professional experience and cloud deployment",

    "Analytics engineer with SQL Python data pipelines and business intelligence",

    "Machine learning developer with Python NLP deep learning and deployment experience",

    "Data analyst with SQL Power BI Python and two years industry experience",

    "Python software engineer with REST APIs SQL testing and deployment experience",

    "Data scientist with statistics Python machine learning and several portfolio projects",

    "Junior developer with Python SQL Git and experience building production applications",

    "Data science intern with Python SQL machine learning and a completed analytics project",

    "Business analyst with SQL Excel analytics and experience working with stakeholders",

    "Machine learning engineer with Python TensorFlow SQL and model deployment experience",

    "Python developer with cloud automation SQL and real world software projects",

    "Data analyst with Python SQL dashboards and experience solving business problems",

    "Software developer with Python backend development SQL and cloud services",

    "Data scientist with Python machine learning NLP and practical research projects",

    "Analytics professional with SQL Python reporting and three years work experience",

    "Python developer with internship experience plus strong Git and deployment projects",

    "Data science graduate with Python SQL statistics and machine learning internship",

    "Machine learning intern with Python deep learning and practical computer vision project",

    "Data analyst with SQL Python Excel and experience creating business dashboards",

    "Python engineer with Django SQL APIs and experience deploying applications",

    "Business intelligence analyst with SQL Power BI Python and industry experience",

    "Data scientist with Python pandas SQL machine learning and cloud experience",

    "Junior software engineer with Python SQL APIs and practical development projects",

    "Data analyst with two years experience Python SQL visualization and reporting",

    "Machine learning graduate with Python deep learning and several completed projects",

    "Python developer with cloud deployment database experience and production projects",

    "Data science intern with SQL Python analytics and strong real world projects",

    "Analytics engineer with SQL Python data pipelines and cloud platform experience",

    "Software engineer with Python backend APIs SQL and deployment experience",

    "Data scientist with machine learning Python SQL and practical business projects",

    "Junior data analyst with Python SQL Power BI and relevant internship experience",
    "Python developer with SQL APIs and experience building practical software projects",
    "Data analyst with Python SQL and dashboard projects plus internship experience"
]


# ============================================================
# 2. REVIEW RESUMES
# ============================================================

review_resumes = [

    "Beginner with Python coursework but no practical project experience",

    "Data science student with basic Python and classroom coursework",

    "Python learner with no professional or project experience",

    "SQL beginner with online courses and limited practical experience",

    "Data science intern with coursework but no completed projects",

    "Machine learning student with basic knowledge and no real world experience",

    "Python developer with only basic training and no practical experience",

    "Analytics learner with online courses and limited hands on experience",

    "Computer science graduate with coursework but no relevant projects",

    "Beginner interested in machine learning with limited practical experience",

    "Python student who completed online courses but has no independent projects",

    "Data analyst beginner with basic SQL knowledge and classroom exercises",

    "Machine learning learner with introductory coursework and no practical applications",

    "Data science student with Python basics but no real world project experience",

    "Python beginner with online certification and limited hands on practice",

    "SQL learner with basic database coursework and no industry experience",

    "Computer science graduate with programming coursework but no relevant projects",

    "Data analytics learner with Excel and basic SQL training but no experience",

    "Machine learning beginner with coursework and simple classroom assignments",

    "Python learner with internship training but no completed technical projects",

    "Data science student with basic Python SQL and theoretical knowledge only",

    "Beginner software developer with online courses and no practical applications",

    "Analytics student with coursework in SQL and Python but limited experience",

    "Python developer with basic training and no professional development experience",

    "Data science intern with classroom knowledge but limited hands on experience",

    "Machine learning student with Python coursework and no real world projects",

    "SQL beginner with certification and basic exercises but no work experience",

    "Computer science student with Python coursework and no substantial projects",

    "Data analyst learner with basic Excel SQL knowledge and no industry experience",

    "Python beginner interested in data science with limited practical skills",

    "Machine learning learner with online courses but no completed projects",

    "Data science graduate with theoretical knowledge but no practical experience",

    "Junior Python learner with coursework and only basic programming exercises",

    "Analytics beginner with online training and limited hands on experience",

    "SQL learner with classroom assignments but no professional analytics experience",

    "Data science student with Python basics and no machine learning projects",

    "Python developer trainee with basic knowledge and no production experience",

    "Machine learning beginner with introductory Python coursework and no projects",

    "Data analyst beginner with online courses and limited practical exposure",

    "Computer science graduate with basic Python and SQL but no relevant experience",

    "Data science learner with internship coursework but no completed real world projects",

    "Python student with certification and basic exercises but no professional experience",

    "Analytics learner with SQL coursework and no business analytics projects",

    "Machine learning student with theoretical understanding and limited practical work",

    "Beginner Python developer with online learning and no deployment experience",

    "Data analyst learner with basic SQL Python and no industry projects",

    "Data science intern with coursework and limited hands on project experience",

    "Python learner with basic programming knowledge and no real world applications",

    "Machine learning beginner with online certification and no practical project",

    "Analytics student with classroom experience but no professional work experience"
]


# ============================================================
# 3. CREATE LABELS AUTOMATICALLY
# ============================================================

shortlist_labels = ["shortlist"] * len(shortlist_resumes)

review_labels = ["review"] * len(review_resumes)


# ============================================================
# 4. COMBINE EVERYTHING
# ============================================================

resume_text = shortlist_resumes + review_resumes

fit_decision = shortlist_labels + review_labels


# ============================================================
# 5. CREATE DATAFRAME
# ============================================================

data = pd.DataFrame({

    "resume_text": resume_text,

    "fit_decision": fit_decision
})


# ============================================================
# 6. CHECK THE DATASET
# ============================================================

print("Number of resumes:", len(data))

print("\nClass distribution:")

print(data["fit_decision"].value_counts())

display(data.head(100))
```

```python
# ============================================================
# 3. SEPARATE INPUT (X) AND TARGET (y)
# ============================================================

X = data["resume_text"]
y = data["fit_decision"]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42,stratify=y)
print("\nTraining resumes:",len(X_train))
print("Testing resumes:",len(X_test))
```

```python
# Create Candidate models
models={
    "Naive Bayes": Pipeline([
        ("tfidf",TfidfVectorizer()),
        ("model",MultinomialNB())
    ]),
    "Logistic Regression": Pipeline([
        ("tfidf",TfidfVectorizer()),
        ("model",LogisticRegression(max_iter=1000))
    ])
}
```

```python
# 6. CROSS-VALIDATION
# ============================================================

cv = StratifiedKFold(

    n_splits=5,

    shuffle=True,

    random_state=42
)


results = {}


for name, model in models.items():

    scores = cross_val_score(

        model,

        X_train,

        y_train,

        cv=cv,

        scoring="f1_macro"
    )

    results[name] = scores.mean()

    print(
        f"\n{name}"
        f"\nF1 scores: {scores.round(3)}"
        f"\nAverage F1: {scores.mean():.3f}"
    )
```

```python
# 7. SELECT THE BEST MODEL
# ============================================================

best_name = max(

    results,

    key=results.get
)


best_model = models[best_name]


print("\n-----------------------------")
print("Best model:", best_name)
print("-----------------------------")
```

```python
best_model.fit(X_train,y_train)
```

```python
predictions=best_model.predict(X_test)
```

```python
# ============================================================
# 10. ACTUAL VS PREDICTED
# ============================================================

comparison = pd.DataFrame({

    "resume": X_test,

    "actual": y_test,

    "predicted": predictions
})


print("\nActual vs Predicted:")
print(comparison.to_string(index=False))
```

```python
# ============================================================
# 11. CHECK WRONG PREDICTIONS
# ============================================================

mistakes = comparison[

    comparison["actual"] != comparison["predicted"]
]


print("\nWrong predictions:")

if len(mistakes) == 0:

    print("No wrong predictions on the test set.")

else:

    print(mistakes.to_string(index=False))
```

```python
# 12. FINAL PERFORMANCE
# ============================================================

accuracy = accuracy_score(

    y_test,

    predictions
)


macro_f1 = f1_score(

    y_test,

    predictions,

    average="macro"
)


print("\nFinal Performance:")
print("Accuracy:", round(accuracy, 3))
print("Macro F1:", round(macro_f1, 3))
```
